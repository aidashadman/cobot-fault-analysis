```python
column_profile = pd.DataFrame({
    "dtype": df.dtypes,
    "non_null": df.notna().sum(),
    "missing": df.isna().sum()
})
column_profile
```

```python
ts = pd.to_datetime(df["Timestamp"].str.strip('"'), utc=True, errors="coerce")
gaps = ts.diff().dt.total_seconds()
print(ts.isna().sum(), ts.min(), ts.max())
print(gaps.describe())
print(gaps.round().value_counts().head())
```

```python
missing_rows = df.isna().any(axis=1)
print(missing_rows.sum())
df[missing_rows].isna().sum()
```

```python
print(df.duplicated().sum(), df["Num"].is_unique)
```

```python
print(df["Robot_ProtectiveStop"].value_counts(dropna=False))
print(df["grip_lost"].value_counts())
print(df["cycle "].nunique(), df["cycle "].min(), df["cycle "].max())
```

### Summary - Data Understanding

- 7,409 rows, 24 columns, one row per second of robot operation.
- `Timestamp` is text and must be converted to datetime.
- `Robot_ProtectiveStop` is float because of missing values.
- Column names to fix: `cycle ` (trailing space), `Temperature_T0`.
- 54 rows contain missing values; in most of them all sensor values are empty.
- No duplicate rows; `Num` is unique.
- One large gap in the timestamps (about 5.2 hours).
- Protective stop: 278 rows (about 3.8%); grip lost: 243 rows (about 3.3%).
